### 1. Data Cleaning and Manipulation on Basic Stats, Advanced Stats and Country Data 

In this section, we will conduct data cleaning and manipulation on the Basic Stats, Advanced Stats, and Bio Data datasets, with the objective of merging them into an unified dataset for future use.

```
# import pertinent libraries for data curation and manipulation
import pandas as pd
import numpy as np

# read datasets for primary dataset of base, advanced NBA Statistics and secondary NBA players' countries dataset
player_info_base = pd.read_csv('Player_Base_Statistics.csv')
player_info_advanced = pd.read_csv('Player_Advanced_Statistics.csv')
player_bio = pd.read_csv('Player_Country.csv')
```

```
# change country names that not USA as international
player_bio.loc[player_bio["COUNTRY"] != "USA", "COUNTRY"] = "international"
```

```
# Concatenate 'COUNTRY' column from player_bio into player_info_base
player_info_base = pd.concat([player_info_base, player_bio["COUNTRY"]], axis=1)
```

```
# display all overlapping columns within player_info_advanced and drop them from player_info_advanced.
# we only keep columns PLAYER_ID','season_id as keys for merging later
overlapping_cols = [
    "PLAYER_NAME",
    "NICKNAME",
    "TEAM_ID",
    "TEAM_ABBREVIATION",
    "AGE",
    "GP",
    "W",
    "L",
    "W_PCT",
    "MIN",
    "FGM",
    "FGA",
    "FG_PCT",
    "GP_RANK",
    "W_RANK",
    "L_RANK",
    "W_PCT_RANK",
    "MIN_RANK",
    "FGM_RANK",
    "FGA_RANK",
    "FG_PCT_RANK",
]

advanced_update = player_info_advanced.drop(overlapping_cols, axis=1)
```

```
# merge player_info_base and advanced_update data into Merged_Stats
player_statistics_country = player_info_base.merge(
    advanced_update, on=["PLAYER_ID", "season_id"], how="inner"
)

# extract the start year of the season_id and convert it as integer type
player_statistics_country["season_id"] = player_statistics_country["season_id"].apply(
    lambda x: x[0:4]
)
player_statistics_country["season_id"] = player_statistics_country["season_id"].astype(
    int
)
```

Now that we have the merged dataset named 'player_statistics_country,' our next step involves manipulating the PLAYER_NAME column. This manipulation is necessary to ensure alignment in format between this dataset and the salary dataset. We will merge these datasets in the future based on player names and years, so it's important to harmonize the format.

```
# eliminate special characters in PLAYER_NAME column
def clean_player_names(df, suffixes_to_remove):
    """The purpose of this function is to eliminate 
    the special characters in the PLAYER_NAME column.
    
    It accepts the following parameters:
        1) the dataframe to be cleaned
        2) a list of special characters to be removed
    """

    for suffix in suffixes_to_remove:
        df['PLAYER_NAME'] = df['PLAYER_NAME'].str.replace(suffix, "", regex=True)
    return df

suffixes_to_remove = ["\."," Jr"," Sr"," III"," II"," IV"]
clean_player_names(player_statistics_country, suffixes_to_remove)
```

```
# eliminate middle names in the PLAYER_NAME column if there is any
stats_first_name = (
    player_statistics_country["PLAYER_NAME"].str.split(" ").apply(lambda x: x[0])
)
stats_last_name = (
    player_statistics_country["PLAYER_NAME"].str.split(" ").apply(lambda x: x[-1])
)

player_statistics_country["PLAYER_NAME"] = stats_first_name + " " + stats_last_name
```

As we observe the data in the 'Player' column above, it contains information about the player's name, position, contract start year, contract end year, and free agent year. We will extract each of these data from the 'Player' column and allocate them to new columns.

```
# extract data from the Player column and assign them to two new columns: names & positions
split_name_yr = df_salary["Player"].str.split("|")

name_position = split_name_yr.apply(lambda x: x[0])

names = name_position.str.split("  ").apply(lambda x: x[0])
positions = name_position.str.split("  ").apply(lambda x: x[1].strip())

print(names[names == 'J.J. Redick'].head(1))
print(names[names == 'Gary Trent Jr.'].head(1))
print(names[names == 'Joe Ingles'].head(1))
print(names[names == 'Frank Mason III'].head(1))

# eliminate special characters in dataframe column names
def clean_player_names(df, suffixes_to_remove):
    """The purpose of this function is to eliminate 
    the special characters in the names column.
    
    It accepts the following parameters:
        1) the dataframe to be cleaned
        2) a list of special characters to be removed
    """
    for suffix in suffixes_to_remove:
        df = df.str.replace(suffix, "", regex=True)
    return df

suffixes_to_remove = ["\."," Jr"," Sr"," III"," II"," IV"]
names = clean_player_names(names, suffixes_to_remove)
```

```
# eliminate middle names and assign player positions to a new column called Positions
first_name = names.str.split(" ").apply(lambda x: x[0].strip())
last_name = names.str.split(" ").apply(lambda x: x[-1].strip())

df_salary["Names"] = first_name + " " + last_name
df_salary["Positions"] = positions
```

```
# check if '.' and 'Jr' in Names column still exists
print(df_salary['Names'][df_salary['Names'] == "JJ Redick"])
print(df_salary['Names'][df_salary['Names'] == "Gary Trent"])
```

```
# extract contract start year, end year, freeagent year data from the Player column and assign them to new columns
start_end_freeagent_yr = split_name_yr.apply(lambda x: x[1])
str_end_free = start_end_freeagent_yr.str.extract(
    "(?P<Start_Year>[\d]*)\-(?P<End_Year>[\d]*)\s\(FA:\s(?P<FreeAgent_Year>.*)\)"
)

# combine str_end_free and df_salary dataframes
df_salary = pd.concat([df_salary, str_end_free], axis=1)
df_salary.head()
```

Next, we will proceed to fill the NaN values in the "Sign Bonus" column with '0'. We will then create a function to clean the salary columns (AVV, Value, Sign Bonus) by removing dollar signs and converting their data types to integers.

```
# fill NaN values with '0' for Sign Bonus column
df_salary["Sign Bonus"] = df_salary["Sign Bonus"].fillna("0")
```

```
# Create a function to transform string values with dollar sign as numeric integer
def extract_and_clean_numeric(value):
    cleaned_values = value.str.replace("[\$,]", "", regex=True)
    return cleaned_values.astype(int)
```

```
# Create a function create new columns to be assigned with data transformed from extract_and_clean_numeric function
def clean_columns(df, column_lst):
    for i in column_lst:
        df[i + "Clean"] = extract_and_clean_numeric(df[i])
```

```
# extract and clean target columns into numeric
clean_columns(df_salary, ["AAV", "Value", "Sign Bonus"])
df_salary.head()
```

It has been identified that 336 rows contain 0 values in either the 'startYear,' 'endYear,' or 'freeAgentYear' columns, and there are 2 rows where all these columns have values of 0. We will create a function to perform data cleaning and populate the appropriate values in these columns.

```
# Create a function to populate the appropriate values in these columns: Start_Year, End_Year, FreeAgent_Year 
# We do not perform manipulation when all three variables are 0 or when all three variables are not 0

def fill_value(row):
    '''The purpose of this function is to fill in most of the Start_Year, End_Year, FreeAgent_Year
    that are 0 values'''
    if row["Start_Year"] == 0 and row["End_Year"] == 0 and row["FreeAgent_Year"] != 0:
        row["End_Year"] = row["FreeAgent_Year"] - 1
        row["Start_Year"] = row["End_Year"] - row["Yrs"] + 1
    if row["Start_Year"] == 0 and row["End_Year"] != 0 and row["FreeAgent_Year"] != 0:
        row["Start_Year"] = row["End_Year"] - row["Yrs"] + 1
    if row["Start_Year"] != 0 and row["End_Year"] == 0 and row["FreeAgent_Year"] != 0:
        row["End_Year"] = row["Start_Year"] + row["Yrs"] - 1
    if row["Start_Year"] != 0 and row["End_Year"] == 0 and row["FreeAgent_Year"] == 0:
        row["End_Year"] = row["Start_Year"] + row["Yrs"] - 1
        row["FreeAgent_Year"] = row["End_Year"] + 1
    if row["Start_Year"] == 0 and row["End_Year"] != 0 and row["FreeAgent_Year"] == 0:
        row["Start_Year"] = row["End_Year"] - row["Yrs"] + 1
        row["FreeAgent_Year"] = row["End_Year"] + 1
    if row["Start_Year"] != 0 and row["End_Year"] != 0 and row["FreeAgent_Year"] == 0:
        row["FreeAgent_Year"] = row["End_Year"] + 1
    return row

#apply fill in function for Start_Year, End_Year, FreeAgent_Year to the df_salary dataframe 
df_salary = df_salary.apply(fill_value, axis=1)
```

We have merged the datasets and saved them to csv - lets do a final check of the dataset to make sure it has mostly non-negative values in the columns and it makes sense.

```
#check to see if values make sense, their type, and that they are generally  negative 
def makes_sense(df):
    '''The purpose of this function is to check if any values in the 
    final merged dataset are negative'''
    for column in df.columns:
        if df[column].dtypes == int or df[column].dtypes == float:
            print(
                'Negative Values Count for '+ column + ': ', 
                df[column][df[column] < 0].count(), 
            )
            print(
                'NaN values for ' + column + ': ', 
                df[column][df[column].isnull()].count(),
                end = '\n\n'
            )
        else:
            print(
                column + ' is not numerical and is type: ', 
                df[column].dtypes
            )
            print(
                'NaN values for ' + column + ': ', 
                df[column][df[column].isnull()].count(),
                end = '\n\n'
            )

#use greater than zero function to check df_final_statistics_salary
makes_sense(df_final_statistics_salary)
```

We can see that the values do make sense, there are <B>no NaN values</B> in columns that are numeric and string columns are all object types. There are a few negative columns such as in ratings, PIE, or plus-minus column but those seem to make sense as those statistics should have negative values in it.
